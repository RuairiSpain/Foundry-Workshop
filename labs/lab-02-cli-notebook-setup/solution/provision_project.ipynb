## Step 2: Create your project

No `account create` here. The hub already exists — this command
attaches a new project to it, the same thing `+ Create Project` did
in the Toolkit in Lab 01.

````python
from provisioning import build_project_create_command, run_az_command

create_cmd = build_project_create_command(
    account_name=ACCOUNT_NAME, resource_group=RESOURCE_GROUP, attendee_id=ATTENDEE_ID
)
print(" ".join(create_cmd))
result = run_az_command(create_cmd)
print(result.stdout)
````

**Expected output:** a JSON object describing your new project, with
`"name": "proj-<your-attendee-id>"`.

## Step 3: Read back your project's endpoint

You'll need this in the next step, and again in Lab 03.

````python
from provisioning import build_project_show_command, parse_project_endpoint

show_cmd = build_project_show_command(
    account_name=ACCOUNT_NAME, resource_group=RESOURCE_GROUP, attendee_id=ATTENDEE_ID
)
result = run_az_command(show_cmd)
endpoint = parse_project_endpoint(result.stdout)
print(f"FOUNDRY_PROJECT_ENDPOINT={endpoint}")
````

Export the endpoint printed above, then reuse Lab 01's verification
